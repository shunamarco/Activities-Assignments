```python
class StoreCart:

    def __init__(self):
        self.products = []

    def add(self, name):
        self.products.append(name)
        print(name, "added.")

    def remove(self, name):
        if name in self.products:
            self.products.remove(name)
            print(name, "removed.")
        else:
            print("Item not found.")

    def display(self):
        if len(self.products) == 0:
            print("Cart is empty.")
        else:
            print("Items in cart:")
            for p in self.products:
                print(p)

    def pay(self):
        print("Checkout items:")
        self.display()
        print("Thank you for buying!")


cart1 = StoreCart()

while True:

    print("\n1 Add")
    print("2 Remove")
    print("3 View")
    print("4 Checkout")
    print("5 Exit")

    option = input("Choose: ")

    if option == "1":
        item = input("Item name: ")
        cart1.add(item)

    elif option == "2":
        item = input("Remove item: ")
        cart1.remove(item)

    elif option == "3":
        cart1.display()

    elif option == "4":
        cart1.pay()
        break

    elif option == "5":
        break

    else:
        print("Invalid input")
```
