```python
def simple_calc():

    print("Choose an operation")
    print("1 - Addition")
    print("2 - Subtraction")
    print("3 - Multiplication")
    print("4 - Division")

    option = input("Pick a number (1/2/3/4): ")

    first = int(input("Enter first value: "))
    second = int(input("Enter second value: "))

    if option == "1":
        result = first + second
        print(first, "+", second, "=", result)

    elif option == "2":
        result = first - second
        print(first, "-", second, "=", result)

    elif option == "3":
        result = first * second
        print(first, "*", second, "=", result)

    elif option == "4":
        if second == 0:
            print("Division by zero is not allowed.")
        else:
            result = first / second
            print(first, "/", second, "=", result)

    else:
        print("Wrong choice")

simple_calc()
```
